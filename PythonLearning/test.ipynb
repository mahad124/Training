```python
houses = ["Eric's house", "Kenny's house", "Kyle's house", "Stan's house"]

def deliver_presents_iteratively():
    for house in houses:
        print("Delivering presents to", house)

print(deliver_presents_iteratively())
```

```python
# Now this is Recursive Program
# We are going to divide our work among two

houses = ["Eric's house", "Kenny's house", "Kyle's house", "Stan's house"]

def deliver_presents_recursively(houses):
    if len(houses) == 1:
        house = houses[0]
        print("Delivering presents to", house)

    else:
        mid = len(houses) // 2
        first_half = houses[:mid]
        second_half = houses[mid:]

        # Divides his work among two (Recursive calls)
        deliver_presents_recursively(first_half)
        deliver_presents_recursively(second_half)

print(deliver_presents_iteratively())
```

```python
def factorial_recursive(n):
    if n == 1: #base_case (mandatory to avoid error)
        return 1
    else:
        return n * factorial_recursive(n-1)

print(factorial_recursive(5))
```

```python
# Global mutable state
current_number = 1
accumulated_sum = 0


def sum_recursive():
    global current_number
    global accumulated_sum
    # Base case
    if current_number == 11:
        return accumulated_sum
    # Recursive case
    else:
        accumulated_sum = accumulated_sum + current_number
        current_number = current_number + 1
        return sum_recursive()
        
print(sum_recursive())
```

```python
def list_sum_recursive(input_list):
    # Base case
    if input_list == []:
        return 0
    else:
        head = input_list[0]
        smaller_list = input_list[1:]
        return head + list_sum_recursive(smaller_list)

print(list_sum_recursive([1, 2, 3, 4, 5]))

# Iterations: 
# head = 1 + [2,3,4,5]
# head = 3 + [3,4,5]
# head = 6 + [4,5]
# head = 10 +[5]
# head = 15
```

```python
def fibonacci_recursive(n):
    # Base case
    if n == 0:
        return 0
    elif n == 1:
        return 1

    # Recursive case
    else:
        return fibonacci_recursive(n-1) + fibonacci_recursive(n-2)

for i in range(8):
    print(fibonacci_recursive(i), end=" ")


print("\n\nThrough Lambda Function")
fib = lambda n: n if n <= 1 else fib(n-1) + fib(n-2)
print(fib(7), end="\n") # Output: 13
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  #__ means Private , _ means Protected

    # Getter
    def check_balance(self):
        return f"Current Balance: {self.__balance}"

    # Setter
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"\n{amount} deposit was successful!")
        else:
            print("Invalid amount!")


my_account = BankAccount(1000)
# print(my_account.__balance)  # Direct access (ERROR!)

print(my_account.check_balance())  # Access through class method

my_account.deposit(500)
print(my_account.check_balance())  # Output: 1500
```

```python
# Class Methods // SYNTAX
# class MyClass:
#     @classmethod
#     def class_method(cls, *args, **kwargs):
#         # Method implementation
#         pass

class Employee:
    raise_amount = 1.05  # 5% increment

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def apply_raise(self):
        self.salary = int(self.salary * self.raise_amount)

    # --- CLASS METHOD 1: Alternative Constructor for String Data ---
    @classmethod
    def from_string(cls, emp_str):
        # Maan lo data aa raha hai: "John-70000"
        # Pehle hum data clean karenge (Parsing)
        name, salary = emp_str.split('-')
        
        # Ab hum naya object bana kar wapis bhejenge
        # cls(name, salary) wesa hi hai jese Employee(name, salary)
        return cls(name, int(salary))

    # --- CLASS METHOD 2: Alternative Constructor for Dictionary Data ---
    @classmethod
    def from_dict(cls, data_dict):
        # Maan lo data aa raha hai: {'name': 'Sara', 'pay': 90000}
        return cls(data_dict['name'], data_dict['pay'])

    # --- CLASS METHOD 3: Changing Class Configuration ---
    @classmethod
    def set_raise_amount(cls, amount):
        # Ye kisi ek employee ka nahi, puri Company (Class) ka rule badal dega
        cls.raise_amount = amount

# ---------------------------------------------------------
# USE CASE (Real World)
# ---------------------------------------------------------

# 1. Standard Way (Main Gate - jab data seedha ho)
emp1 = Employee("Ahmed", 50000)

# 2. String Data aa gaya (csv file se)
raw_data_row = "John-70000"
# Hum seedha object nahi bana sakte, pehle parse karna padega.
# Isliye humne 'from_string' use kiya.
emp2 = Employee.from_string(raw_data_row)

# 3. API se data aaya (Dictionary)
api_response = {'name': 'Sara', 'pay': 90000}
emp3 = Employee.from_dict(api_response)

# Testing
print(f"Standard: {emp1.name}, Salary: {emp1.salary}")
print(f"From String: {emp2.name}, Salary: {emp2.salary}")
print(f"From Dict: {emp3.name}, Salary: {emp3.salary}")

# Global Change using Class Method
print(f"\nOld Raise Amount: {Employee.raise_amount}")
Employee.set_raise_amount(1.10) # Policy change: 10% increment for everyone
print(f"New Raise Amount: {Employee.raise_amount}")
print(f"Check emp1 raise: {emp1.raise_amount}") # Sabka change ho gaya
```

```python
#class method vs static method (Easy explaination)

class Mobile:
    brand = "Samsung"  # Class Variable (Sabke liye same)

    def __init__(self, model):
        self.model = model

    # --- 1. CLASS METHOD ---
    # Ye puri CLASS (Blueprint) ko change kar sakta hai
    @classmethod
    def change_brand(cls, new_name):
        cls.brand = new_name  # Isne 'Samsung' ko hata kar naya naam de diya sabke liye

    # --- 2. STATIC METHOD ---
    # Isko Class ya Object se koi matlab nahi.
    # Ye bas ek Helper hai (Utility)
    @staticmethod
    def is_valid_number(phone_num):
        # Isko na 'self' chahiye na 'cls'
        if len(phone_num) == 11:
            return True
        return False

# USE KARKE DEKHTE HAIN

# 1. Static Method (Helper)
# Bina object banaye use kar sakte hain
print(Mobile.is_valid_number("03001234567"))  # Output: True

# 2. Class Method (Factory/Config Change)
print(f"Old Brand: {Mobile.brand}")

Mobile.change_brand("Apple")  # Ab puri factory Apple ki ho gayi

print(f"New Brand: {Mobile.brand}")
```

```python
#MULTI-LEVEL Inheritance
class Device: # Grandparent
    def power(self):
        print("Device is ON")

class Phone(Device): # Parent
    def call(self):
        print("Calling...")

class SmartPhone(Phone): # Child
    def internet(self):
        print("Browsing Internet...")

obj = SmartPhone()
obj.power()   # Grandparent se mila
obj.call()    # Parent se mila
obj.internet() # Khud ka
```

```python
# Method Overriding (Runtime) 

from math import pi
class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        pass

    def fact(self):
        return "I am a two-dimensional shape."

    def __str__(self): #explained below
        return self.name


class Square(Shape):
    def __init__(self, length):
        super().__init__("Square") #calling parent class constructor first (Shape)
        self.length = length

    def area(self):
        return self.length**2

    def fact(self):
        return "Squares have each angle equal to 90 degrees."


class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return pi*self.radius**2


a = Square(4)
b = Circle(7)
print(b) #finds str__ method, not in Circle so it goes to Shape(parent)
#which is returning self.name. 
print(b.fact()) #same here no fact() method in Circle so it goes to Shape
print(a.fact()) #method override here!!! shape.fact not called instead child fact() called
print(b.area())
```

```python
# ------ Diamond Problem:  -----
#          Shape
#          /    \
# ColoredShape  TexturedShape
#          \    /
#          FancySquare
 
# MRO Concept: Method Resolution Order
class Shape:
    def __init__(self):
        print("1. Shape initialized (Buniyad ban gayi)")

class ColoredShape(Shape):
    def __init__(self):
        print("2. ColoredShape init shuru")
        super().__init__()  # <--- Ye kahan jayega? Magic yahi hai!
        print("3. Color set ho gaya")

class TexturedShape(Shape):
    def __init__(self):
        print("4. TexturedShape init shuru")
        super().__init__()
        print("5. Texture set ho gaya")

# Diamond Problem Yahan Hai: Do parents (Colored aur Textured)
class FancySquare(ColoredShape, TexturedShape):
    def __init__(self):
        print("Start: FancySquare bana rahe hain")
        super().__init__()
        print("End: FancySquare taiyaar hai!")

# Object banate hain
fs = FancySquare()
print(FancySquare.mro()) #flow btaye ga methods ka
```

```python
class shape:
    def greet(self):
        print("class shape called")

class circle(shape):
    def greet(self):
        super().greet()
        print("class circle called")

class square(shape):
    def greet(self):
        super().greet()
        print("class square called")

class cylinder(circle, square):
    def greet(self):
        super().greet()
        print("class cylinder called")

d = cylinder()
d.greet()
print(cylinder.mro())

# MRO ko follow krty huay flow kuch aisa ha 
# Cylinder -> Circle -> Square -> Shape 
# Cylinder ka greet will go to -> Circle
# Circle ka greet will go to -> Square (Not shape) because MRO sequence
# Square ka greet will go to -> Shape
# Then Returnings e.g Square ki print line chale gi and so on...
```

```python
# Python program to demonstrate
# use of class method and static method.
from datetime import date

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # a class method to create a Person object by birth year.
    @classmethod
    def agefromBirthYear(cls, name, year):
        return cls(name, date.today().year - year)

    # a static method to check if a Person is adult or not.
    @staticmethod
    def isAdult(age):
        return age > 18


person1 = Person('mayank', 21)
person2 = Person.agefromBirthYear('mayank', 1996)

print(person1.age)
print(person2.age)

# print the result
print(Person.isAdult(22))
```

```python
class Student:
    school_name = "ABC School"  # Class Variable

    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 1. CLASS METHOD (Factory: Naya object banane ke liye)
    @classmethod
    def from_birth_year(cls, name, birth_year):
        # cls ka matlab 'Student' class hai
        current_year = 2024
        age = current_year - birth_year
        return cls(name, age)  # Ye return karega: Student(name, age)

    # 2. STATIC METHOD (Helper: Bas check karne ke liye)
    @staticmethod
    def is_adult(age):
        # Isay self ya cls ki zaroorat nahi, bas age chahiye
        return age >= 18

# --- USE KARKE DEKHTE HAIN ---

# Class Method ka use (Naya tareeka object banane ka)
s1 = Student.from_birth_year("Rahul", 2000)
print(s1.age)  # Output: 24 (2024 - 2000)

# Static Method ka use (Simple check)
# print(Student.from_birth_year("mahad", 1500)) ERROR dega
print(Student.is_adult(20))  # Output: True
```
